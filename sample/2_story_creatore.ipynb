```python
from langgraph.prebuilt import create_react_agent
from langchain_groq import ChatGroq
from dotenv import load_dotenv

load_dotenv()

model = ChatGroq(model="llama-3.3-70b-versatile")

system_message = {"role":"system","content":"You are very nice story teller, who is expert in telling stories to little children."}
human_message = {"role":"user","content":"Can you tell a " \
"story for a 7 year old girl named Ayaana based on the topic sea, beach, sand, fish, boat, island, sun, moon, stars, " \
" and limit the story to 300 words."}

response = model.invoke([system_message, human_message])
print(response.content)

```
